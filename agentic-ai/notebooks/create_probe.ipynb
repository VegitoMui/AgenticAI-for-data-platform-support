```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service import jobs
from databricks.sdk.service.workspace import ImportFormat, Language

w = WorkspaceClient()
me = w.current_user.me().user_name
path = f"/Users/{me}/agentic_ai_probe_v3"

w.workspace.upload(
    path=path,
    content=(
        b"# Databricks notebook source\n"
        b"spark.table('databricks_ws.agentic_ai_dev.incidentz').count()\n"
    ),
    format=ImportFormat.SOURCE, language=Language.PYTHON, overwrite=True,
)
job = w.jobs.create(
    name="agentic-ai probe v3 (missing table)",
    tasks=[jobs.Task(task_key="read_missing_table",
                     notebook_task=jobs.NotebookTask(notebook_path=path))],
)
run = w.jobs.run_now(job_id=job.job_id)
try:
    run.result()
except Exception as e:
    print("failed as expected:", str(e)[:120])
print("job_id:", job.job_id, "| run_id:", run.run_id)
```

```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service import jobs
from databricks.sdk.service.workspace import ImportFormat, Language

w = WorkspaceClient()
me = w.current_user.me().user_name
path = f"/Users/{me}/agentic_ai_probe_v5"

w.workspace.upload(
    path=path,
    content=(
        b"# Databricks notebook source\n"
        b"spark.sql('SELECT custmer_id FROM databricks_ws.agentic_ai_dev.incidents').show()\n"
    ),
    format=ImportFormat.SOURCE, language=Language.PYTHON, overwrite=True,
)
job = w.jobs.create(
    name="agentic-ai probe v5 (unresolved column)",
    tasks=[jobs.Task(task_key="query_bad_column",
                     notebook_task=jobs.NotebookTask(notebook_path=path))],
)
run = w.jobs.run_now(job_id=job.job_id)
try:
    run.result()
except Exception as e:
    print("failed as expected:", str(e)[:120])
print("job_id:", job.job_id)
```

```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()
for name in ("agentic-ai probe v3 (missing table)", "agentic-ai probe v5 (unresolved column)"):
    job = next(j for j in w.jobs.list() if j.settings.name == name)
    print(name, "-> run", w.jobs.run_now(job_id=job.job_id).run_id)
```
